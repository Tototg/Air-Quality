```python
candidate_sensors = {
    "Palermo": 15012777,
    "Adrogué": 9224161,
    "Dock Sud (AirGradient)": 9224148,
    "La Matanza": 7696815,
    "La Plata / UNLP": 7466355,
    "Neuquén - Plottier": 16246050,
}

def check_recent_data(name, sensor_id, days=30):
    end_date = datetime.now(timezone.utc)
    start_date = end_date - timedelta(days=days)

    response = client.measurements.list(
        sensors_id=sensor_id,
        data="hours",
        datetime_from=start_date,
        datetime_to=end_date,
        limit=1000
    )

    records = response.results

    if not records:
        print(f"{name}: sin datos en los últimos {days} días")
        return

    first_date = records[0].period.datetime_from.local
    last_date = records[-1].period.datetime_from.local

    print(
        f"{name}: {len(records)} registros | "
        f"desde {first_date} hasta {last_date}"
    )
```

```python
for station_name, sensor_id in candidate_sensors.items():
    check_recent_data(station_name,sensor_id)
```

```python
rows = []

for measurement in matanza_data_hourly.results:
    rows.append({
        "measured_at_utc" : measurement.period.datetime_from.utc,
        "measured_at_local": measurement.period.datetime_from.local,
        "pm25_ug_m3": measurement.value,
        "coverage_pct": measurement.coverage.percent_complete,
        }
    )
    
pm25_df = pd.DataFrame(rows)
pm25_df.head()
```

```python
historical_matanza_data = client.measurements.list(sensors_id=7696815,
                                                   data="days",
                                                   date_from="2022-01-01",
                                                   date_to="2026-09-08",
                                                   limit=1000)

historical_matanza_data.results[0]
```

```python
historical_rows = []

for measurement in historical_matanza_data.results:
    historical_rows.append(
        {
            "period_start_utc": measurement.period.datetime_from.utc,
            "period_start_local": measurement.period.datetime_from.local,
            "pm25_ug_m3": measurement.value,
            "coverage_pct": measurement.coverage.percent_complete,
        }
    )

historical_daily_df = pd.DataFrame(historical_rows)

historical_daily_df["period_start_utc"] = pd.to_datetime(
    historical_daily_df["period_start_utc"],
    utc=True
)

historical_daily_df = historical_daily_df.sort_values(
    "period_start_utc"
).reset_index(drop=True)
```

```python
print("Primera fecha:", historical_daily_df["period_start_utc"].min())
print("Última fecha:", historical_daily_df["period_start_utc"].max())
print("Días con datos:", len(historical_daily_df))
print("\nCobertura diaria:")
print(historical_daily_df["coverage_pct"].describe())
```

```python
historical_daily_df["gap_days"] = (
    historical_daily_df["period_start_utc"]
    .diff()
    .dt.total_seconds()
    .div(86_400)
)

gaps = historical_daily_df.loc[
    historical_daily_df["gap_days"] > 1,
    ["period_start_utc", "gap_days"]
]

gaps.head(20)
```
